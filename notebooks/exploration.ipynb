# Crop Yield Prediction - Exploratory Data Analysis (EDA)

This notebook performs exploratory data analysis on the real FAOSTAT & World Bank global crop yield dataset (`data/raw/crop_yield.csv`).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load raw FAO & World Bank dataset
df = pd.read_csv("../data/raw/crop_yield.csv")
if "Unnamed: 0" in df.columns:
    df = df.drop(columns=["Unnamed: 0"])

# Convert target to tons/hectare
df["yield_tons_per_ha"] = df["hg/ha_yield"] / 10000.0

print(f"Dataset Shape: {df.shape}")
df.head()
```

## 1. Dataset Structure & Missing Values

```python
print(df.info())
print("\nMissing values per column:")
print(df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
```

## 2. Statistical Summary of Numerical Features & Target

```python
df.describe()
```

## 3. Crop & Country Distribution

```python
print("Top 10 Countries by Record Count:")
print(df["Area"].value_counts().head(10))
print("\nCrop Record Distribution (Item):")
print(df["Item"].value_counts())
```

## 4. Yield Distribution by Crop Type

```python
print("Yield Statistics (tons/ha) by Crop:")
print(df.groupby("Item")["yield_tons_per_ha"].describe())
```

## 5. Key EDA Findings
- **Total Records**: 28,242 real country-crop observations across 101 countries (1990-2013).
- **Target Variable**: Yield averages ~7.70 tons/ha across crops, ranging from ~0.01 tons/ha (low yield regions) up to 50.14 tons/ha (high yield crops like Potatoes/Cassava).
- **Features**: 2 categorical (`Area`, `Item`) and 4 numerical (`Year`, `average_rain_fall_mm_per_year`, `pesticides_tonnes`, `avg_temp`).
- Dataset is complete with 0 missing values and clean data structure.
