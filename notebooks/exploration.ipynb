# Crop Yield Prediction - Exploratory Data Analysis (EDA)

This notebook performs exploratory data analysis on the crop yield dataset to understand distributions, relationships between features, and target variable characteristics.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load raw dataset
df = pd.read_csv("../data/raw/crop_yield.csv")
print(f"Dataset Shape: {df.shape}")
df.head()
```

## 1. Dataset Overview & Data Types

```python
print(df.info())
print("\nMissing values per column:")
print(df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
```

## 2. Statistical Summary

```python
df.describe()
```

## 3. Categorical Distributions (Crop & Soil Types)

```python
print("Crop Value Counts:")
print(df["crop"].value_counts())
print("\nSoil Type Value Counts:")
print(df["soil_type"].value_counts())
```

## 4. Target Distribution (Yield)

```python
print(df.groupby("crop")["yield"].describe())
```

## 5. Summary Observations
- 1500 complete records with 0 missing values.
- Features include numeric agricultural metrics (area, rainfall, temperature, humidity, fertilizer) and categorical variables (crop, soil_type).
- Yield varies drastically by crop (Sugarcane highest tons/ha, Cotton/Grains lower).
- Ready for preprocessing and ML model pipeline.
